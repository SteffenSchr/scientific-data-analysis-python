```python
#########################################################
##################### Start of the code ##################
#########################################################

########## Path of the folder with the measurement files ##########
path = '../data/Batt15/raw/rate-test' #The input data is in the folder given by path
print('-------------------')
print('The following Path is selected:', path)
print('-------------------')
```

```python
########## Handling of the raw data ##########
#############################################

########## List all files in a directory using scandir() ##########

########## Walking a directory tree and printing the names of the directories and files ########## (only for an overview)
#print('-------------------')
#print('The Following folder and File Structure in the given path is found')
#for foldername, subfolders, filenames in os.walk(path):
#    print(f'Found directory: {foldername}')
#    for file_name in filenames:
#        print(file_name)
#print('-------------------')


########## All files and Directories given as a Dataframe ##########
dataframe_of_allfiles = pd.DataFrame(list(os.walk(path)))  # First column: directory/path ; second column: directories in the given directory/path ; third column: files in the given directory/path
                                              # os.walk('path') returns triples (dirpath, dirnames, files) -> list(os.walk('.')) puts these triples in a list


########## Sort the found files by name ##########
FilterbyName1 = 'Batt'
FilterbyName3 = '_CP_'
FilterbyName4 = '_OCV_'
FilterbyName5 = 'Cdis'
FilterbyName6 = 'Cch'

print('-------------------')
print('Only Files with the following Name are considered:', FilterbyName1)
print('The Files with the Name ', FilterbyName1, 'is split in three groups by the names:', FilterbyName3, 'and', FilterbyName4)
print('These two groups are split again in two goups by the names:', FilterbyName5, 'and', FilterbyName6)
print('-------------------')

list_of_filenames = dataframe_of_allfiles[2][0]

########## Sorting the battery files ##########
list_of_BatteryMeasurements = sorted([k for k in list_of_filenames if FilterbyName1 in k])

########## Filter by the wanted file format ##########
Format = 'mpt'
print('-------------------')
print('Only Files in the Lists with the following Format are considered:', Format)
print('-------------------')

for i in range(len(list_of_BatteryMeasurements)):
    list_of_BatteryMeasurements = [i for i in list_of_BatteryMeasurements if i.endswith(Format)]

########## Sort the OCV steps before the CP step ##########
#list_of_OCV_before_CP_step = sorted([k for k in list_of_BatteryMeasurements if FilterbyName2 in k])
#print(list_of_OCV)

#list_of_OCV_before_CP_step_charge = sorted([k for k in list_of_OCV_before_CP_step if FilterbyName6 in k])
#print(list_of_OCV_before_CP_step_charge)

#list_of_OCV_before_CP_step_discharge = sorted([k for k in list_of_OCV_before_CP_step if FilterbyName5 in k])
#print(list_of_OCV_before_CP_step_discharge)

########## Sort the CP steps ##########
list_of_CP = sorted([k for k in list_of_BatteryMeasurements if FilterbyName3 in k])
#print(list_of_PEIS)

list_of_CP_charge = sorted([k for k in list_of_CP if FilterbyName6 in k])
#print(list_of_CP_charge)

list_of_CP_discharge = sorted([k for k in list_of_CP if FilterbyName5 in k])
#print(list_of_CP_discharge)

########## Sort the OCV steps after the CP step ##########
list_of_OCV_after_CP_step = sorted([k for k in list_of_BatteryMeasurements if FilterbyName4 in k])

list_of_OCV_after_CP_step_charge = sorted([k for k in list_of_OCV_after_CP_step if FilterbyName6 in k])
#print(list_of_OCV_after_CP_step_charge)

list_of_OCV_after_CP_step_discharge = sorted([k for k in list_of_OCV_after_CP_step if FilterbyName5 in k])
#print(list_of_OCV_after_CP_step_discharge)


########## Join the path of the files with the file names ########
#for i in range(len(list_of_allfiles)):
#    for j in range(len(list_of_allfiles[i])):
#      list_of_allfiles[i][j] = (dataframe_of_allfiles.iloc[i].iloc[0] + '/' + list_of_allfiles[i][j]) # joins the path (in 'allfiles' column 0) with the file name (in 'lists')
```

```python
########## Read the files into dataframes ##########

##### read the CP steps
list_of_dataframes_of_CP_charge = Read_Biologic_Mpt_Files(path, list_of_CP_charge, '.')
list_of_dataframes_of_CP_discharge = Read_Biologic_Mpt_Files(path, list_of_CP_discharge, '.')

##### read the OCV steps after the CP step
list_of_dataframes_of_OCV_after_CP_step_charge = Read_Biologic_Mpt_Files(path, list_of_OCV_after_CP_step_charge, '.')
list_of_dataframes_of_OCV_after_CP_step_discharge = Read_Biologic_Mpt_Files(path, list_of_OCV_after_CP_step_discharge, '.')
```
