```python
########################################################
###################### Importierte Packete #############
########################################################

##### Visualisieren
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import rc                                                   # To use a LaTeX font

##### Calculations and mathematical functions
import numpy as np

##### Data Management
import pandas as pd

##### Scientific functions
import scipy
from scipy import stats
import scipy.constants as sc
from scipy import ndimage
import scipy.signal

from mpl_toolkits import mplot3d                                            # Um 3D Plots zu machen

import os

##### Saving/packing of data
import pickle
```

```python
#########################################################
##################### Start of the code ##################
#########################################################

########## Path of the folder with the measurement files ##########
path = './data/raw' #The input data is in the folder given by path
print('-------------------')
print('The following Path is selected:', path)
print('-------------------')
```

```python
########## Handling of the raw data ##########
#############################################

########## List all files in a directory using scandir() ##########

########## Walking a directory tree and printing the names of the directories and files ########## (only for an overview)
#print('-------------------')
#print('The Following folder and File Structure in the given path is found')
#for foldername, subfolders, filenames in os.walk(path):
#    print(f'Found directory: {foldername}')
#    for file_name in filenames:
#        print(file_name)
#print('-------------------')


########## All files and Directories given as a Dataframe ##########
dataframe_of_allfiles = pd.DataFrame(list(os.walk(path)))  # First column: directory/path ; second column: directories in the given directory/path ; third column: files in the given directory/path
                                                            # os.walk('path') returns triples (dirpath, dirnames, files) -> list(os.walk('.')) puts these triples in a list

######### Returns the files found in the data folder ########
#print('Following Files are found in the Folder, in which the Python File is located:', dataframe_of_allfiles[2][0])    

     
########## Sort the found files by name ##########
FilterbyName1 = 'Batt'
#FilterbyName2 = '_01_OCV_'
FilterbyName3 = '_PEIS_C'
FilterbyName4 = 'Cdis'
FilterbyName5 = 'Ccharge'

print('-------------------')
print('Only Files with the following Name are considered:', FilterbyName1)
print('The Files with the Name ', FilterbyName1, 'is split in one groupe by the name:', FilterbyName3 )
print('These two groups are split again in two goups by the names:', FilterbyName4, 'and', FilterbyName5)
print('-------------------')

list_of_filenames = dataframe_of_allfiles[2][0]

list_of_BatteryMeasurements = sorted([k for k in list_of_filenames if FilterbyName1 in k])

########## Filter by the wanted file format ##########
Format = 'mpt'
print('-------------------')
print('Only Files in the Lists with the following Format are considered:', Format)
print('-------------------')

for i in range(len(list_of_BatteryMeasurements)):
    list_of_BatteryMeasurements = [i for i in list_of_BatteryMeasurements if i.endswith(Format)]

#list_of_OCV = sorted([k for k in list_of_BatteryMeasurements if FilterbyName2 in k])
#print(list_of_OCV)

#list_of_OCV_charge = sorted([k for k in list_of_OCV if FilterbyName5 in k])
#print(list_of_OCV_charge)

#list_of_OCV_discharge = sorted([k for k in list_of_OCV if FilterbyName4 in k])
#print(list_of_OCV_discharge)

list_of_PEIS = sorted([k for k in list_of_BatteryMeasurements if FilterbyName3 in k])
print(len(list_of_PEIS))

list_of_PEIS_charge = sorted([k for k in list_of_PEIS if FilterbyName5 in k])
print(len(list_of_PEIS_charge))

list_of_PEIS_discharge = sorted([k for k in list_of_PEIS if FilterbyName4 in k])
print(len(list_of_PEIS_discharge))


########## Join the path of the files with the file names ########
#for i in range(len(list_of_allfiles)):
#    for j in range(len(list_of_allfiles[i])):
#      list_of_allfiles[i][j] = (dataframe_of_allfiles.iloc[i].iloc[0] + '/' + list_of_allfiles[i][j]) # joins the path (in 'allfiles' column 0) with the file name (in 'lists')
```

```python
########## Read the files into dataframes ##########
list_of_dataframes_of_PEIS_charge = []
for i in range(len(list_of_PEIS_charge)):
    count = -2
    with open(os.path.join(path, list_of_PEIS_charge[i]), 'r', encoding='unicode_escape') as infile:
        for line in infile:
            count += 1
            if line.startswith("freq"):
                break
    file = pd.read_csv(os.path.join(path, list_of_PEIS_charge[i]), sep='\t', encoding='unicode_escape', skiprows=count, decimal='.', dtype=np.float64)
    list_of_dataframes_of_PEIS_charge.append(file)
print(len(list_of_dataframes_of_PEIS_charge)) 

list_of_dataframes_of_PEIS_discharge = []
for i in range(len(list_of_PEIS_discharge)):
    count = -2
    with open(os.path.join(path, list_of_PEIS_discharge[i]), 'r', encoding='unicode_escape') as infile:
        for line in infile:
            count += 1
            if line.startswith("freq"):
                break
    file = pd.read_csv(os.path.join(path, list_of_PEIS_discharge[i]), sep='\t', encoding='unicode_escape', skiprows=count, decimal='.', dtype=np.float64)
    list_of_dataframes_of_PEIS_discharge.append(file)
print(len(list_of_dataframes_of_PEIS_discharge)) 



#list_of_dataframes_of_PEIS_charge = []
#for i in range(len(list_of_PEIS_charge)):
#    file = pd.read_csv(os.path.join(path, list_of_PEIS_charge[i]), sep='\t', encoding='unicode_escape', header=65, decimal='.', dtype=np.float64)
#    list_of_dataframes_of_PEIS_charge.append(file)

#print(list_of_dataframes_of_PEIS_charge)    
    
#list_of_dataframes_of_PEIS_discharge = []
#for i in range(len(list_of_PEIS_discharge)):    
#    file = pd.read_csv(os.path.join(path, list_of_PEIS_discharge[i]), sep='\t', encoding='unicode_escape', header=65, decimal='.', dtype=np.float64)
#    list_of_dataframes_of_PEIS_discharge.append(file)
```
