```python
import re

# Extract for every sample the indices in the dataframe and save it in a dictonary
str_list = data[0]['Unnamed: 0']
# define the two substring in which between is the relevant substring which should be extracted and matched with a index
substring_start = r'Spot\d{2}_\d{2}_'
substring_end = r'_#'
sample_names = {} # dictionary like this: {'L9': [02, 03, 04], [0,1,2,3]} = spot idx list and list_idx lsit xDDD

idx = 0

for string in str_list:
    #get sample name
    str_split = string.split(substring_end)
    #print(str_split)
    sample_name = str_split[-2].split(substring_start)[-1]
    #print(sample_name)
    #add sample to dict if not already in it
    if sample_name not in sample_names:
        sample_names[sample_name] = ([idx])
    else:
        sample_names[sample_name].append(idx)
    idx += 1
    
print(sample_names)
print('|')
print(sample_names.keys())
```

```python
import re

# Your dataframe source
str_list = data[0]['Unnamed: 0']

# Define patterns using Regex
# \d{2} matches exactly 2 digits
substring_start = r'Spot\d{2}_\d{2}_'
substring_end = r'_#'

sample_names = {}

for idx, string in enumerate(str_list):
    # We create a combined pattern: StartPattern + (CaptureGroup) + EndPattern
    # re.escape is used on the end marker to handle the '#' safely
    full_pattern = rf"{substring_start}(.*?){re.escape(substring_end)}"
    
    match = re.search(full_pattern, string)
    
    if match:
        # match.group(1) contains exactly what was inside the (.*?)
        sample_name = match.group(1)
        
        # Add to dictionary logic
        if sample_name not in sample_names:
            sample_names[sample_name] = [idx]
        else:
            sample_names[sample_name].append(idx)
    else:
        # Optional: handle strings that don't match the pattern
        pass

print(sample_names)
print('|')
print(list(sample_names.keys()))
```

## PCA Analysis

```python
### PCA Analysis

from sklearn.decomposition import PCA

### User Input
number_of_calculated_components = 5
print('Number of Principal Components is:', number_of_calculated_components)

print('|')

### Define the Name of the Components
list_of_component_names = []
for i in range(number_of_calculated_components):
    name = 'principal component %s' %(i+1)
    
    list_of_component_names.append(name)
    
# Set options for the PCA
pca = PCA(n_components=number_of_calculated_components)

print('Performing PCA')
print('.\n.\n.')
# Fit the PCA to the Data
principal_components_fitted = pca.fit(data_array)
print('PCA finished')
```

### Explained Variance

```python
import numpy as np

nr_pcs = principal_components_fitted.components_.shape[0]
print('Number of Components of the PCA is: %s' %(nr_pcs))

exp_variance = principal_components_fitted.explained_variance_ratio_
print('Explained Variance of the Data by the Components is: %s in Ratios' %(exp_variance))
```
